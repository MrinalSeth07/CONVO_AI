# Stemming:
Is the process of converting the stem word to its base form that affixes to the prefix and the suffix to the roots of the word that is know as the lemma



Nltk has both stemming and the lemmatization but the Spacy has only the Lemmatization

```python
from nltk.stem import PorterStemmer
```

```python
#POSTER STEMMER
stemming = PorterStemmer()
```

```python
words=["eating","ate","eaten","history","congratulatIons"]
```

```python
for word in words:
    print(word+"----->"+stemming.stem(word))
```

As we can see it implies stemming to thewords where it is not required as we can see above in the congratulations and the history
Stemming follows the rule base approach :

    Example (Porter Stemmer rules):
    
    - ing → "" if word length > 4
    
    - ies → y
    
    - ed → "" if word length > 3

# Lancaster Stemming :

It uses about 120 rules that define how to strip or replace word endings.

Each rule says:

 - What suffix to look for

 - Whether to delete or replace it

 - Conditions (like minimum word length)

The algorithm repeatedly applies these rules until no more rules apply.


More aggresive compared to porter useful when need the more normalized words but not when need for the dictionary words


```python
#LANCASTER STEMMING ALGO
from nltk.stem import  LancasterStemmer
```

```python
lancaster = LancasterStemmer()
```

```python
for word in words:
    print(word+" -----> "+lancaster.stem(word))
```

# Snowball Stemmer
Like Porter and Lancaster, it is rule-based.

But compared to Porter:

More consistent and cleaner rules.

Handles edge cases better.

Easier to extend for other languages.

Compared to Lancaster:

Less aggressive (doesn’t over-chop words too much).

Produces more linguistically reasonable stems.


```python
from nltk.stem import SnowballStemmer
```

```python
stemmer=SnowballStemmer('english',ignore_stopwords=True)
for word in words:
    print(word + " -----> " +stemmer.stem(word))
    
```

# Wordnet Lemmitization

this is like stemming . The output we get after the lemmitization is called 'lemma' which is root word rather then root stem.After lemmetization we will be getting the valid word that means the same thing .

NLTK provides WorldNetLemmitizater class is a thin wrapperaround the wordnet corpus. This class use the morphy() function to WordNet Corpus Reader class to find a lemma

```python
import nltk

nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
lemm=WordNetLemmatizer()
```

```python
#POS 
# noun-n
# adjective-a
# adverb-r
# verb-v
for word in words:
  print(word+' ---> '+lemm.lemmatize(word))  #add the pos along wiht word to make it act by taking it as the pos tagging
```
